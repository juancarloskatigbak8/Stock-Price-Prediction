```python
import pandas as pd
import os

# Get the current working directory (fix for Jupyter Notebook)
script_dir = os.getcwd()

# File paths
csv_file = os.path.join(script_dir, 'all_stocks_5yr.csv')
parquet_file = os.path.join(script_dir, 'all_stocks_5yr.parquet')

# Check if the CSV file exists
if not os.path.exists(csv_file):
    print(f"Error: {csv_file} not found. Please place the CSV file in the same folder as this notebook.")
else:
    # Read CSV
    df = pd.read_csv(csv_file)
    print(df.head())

    # Convert to Parquet
    df.to_parquet(parquet_file, engine='pyarrow', compression='snappy')
    print(f"Parquet file created: {parquet_file}")

    # Read back Parquet to confirm
    df_parquet = pd.read_parquet(parquet_file, engine='pyarrow')
    print(df_parquet.head())
```

```python
%timeit pd.read_csv(csv_file)

%timeit pd.read_parquet(parquet_file, engine='pyarrow')
```

```python
df_10x = pd.concat([df] * 10, ignore_index=True)

df_10x.to_csv('all_stocks_5yr_10x.csv', index=False)
df_10x.to_parquet('all_stocks_5yr_10x.parquet', engine='pyarrow', compression='snappy')
print("10x dataset created!")
```

```python
%timeit pd.read_csv('all_stocks_5yr_10x.csv')

%timeit pd.read_parquet('all_stocks_5yr_10x.parquet', engine='pyarrow')
```
